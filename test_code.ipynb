```python
student_name = ['serria', 'susan', 'sylvia' 'peter', 'mike', 'ray']

result = []
search_string = "s"

for a in student_name:
    if a.lower().startswith(search_string):
        result.append(a)

if result:
    print("search returned:", )
    for w in result:
        print(w)
else:
    print("Search result didn't return", search_string)
    
```
